<b>Step-by-Step Method Description</b>

<ul>
<li><b>Step 1: Epoch Segmentation of the Document Stream</b><br>
<ul>
<li><b>Description of what happens in this step</b>: The incoming stream of documents is partitioned into discrete time slices called epochs (e.g., days). Each epoch contains a set of documents.</li>
<li><b>Reference</b>: Section 2 (Problem Setup); Figure 1.</li>
<li><b>Purpose of this step</b>: Creates the temporal scaffolding that the model needs to track topic changes over time.</li>
</ul>
</li><br>

<li><b>Step 2: Maintaining Multiscale Word Distributions from Previous Epochs</b><br>
<ul>
<li><b>Description</b>: For each topic and each timescale, the model maintains estimated multiscale word distributions using exponential smoothing with varying decay rates.</li>
<li><b>Reference</b>: Section 3.1, Equations (5)–(6).</li>
<li><b>Purpose</b>: Provides the model with temporal memory at multiple granularities, allowing it to track both fast bursty changes and stable long-term trends.</li>
</ul>
</li><br>

<li><b>Step 3: Constructing the Dirichlet Prior</b><br>
<ul>
<li><b>Description</b>: Before generating documents at the current epoch, the model constructs a Dirichlet prior for each topic's word distribution using a weighted combination of the multiscale distributions from previous epochs.</li>
<li><b>Reference</b>: Section 3.1, Equations (1)–(3).</li>
<li><b>Purpose</b>: Links past topic behaviour to the current epoch, biasing current word distributions toward historical patterns.</li>
</ul>
</li><br>

<li><b>Step 4: Generative Process for Documents</b><br>
<ul>
<li><b>Description</b>: For each epoch, words are generated by drawing topic proportions from a Dirichlet distribution, and then for each word drawing a topic and drawing the word from the topic's distribution (which is sampled from the Dirichlet prior created in Step 3).</li>
<li><b>Reference</b>: Section 3.1 (Generative Model).</li>
<li><b>Purpose</b>: Defines the probabilistic data-generation story incorporating temporal dynamics.</li>
</ul>
</li><br>

<li><b>Step 5: E-step — Collapsed Gibbs Sampling</b><br>
<ul>
<li><b>Description</b>: The model infers the latent topic assignments using collapsed Gibbs sampling, where the conditional probability combines the current epoch's word counts with the multiscale prior.</li>
<li><b>Reference</b>: Section 3.2, Equations (7)–(8).</li>
<li><b>Purpose</b>: Discovers the topic of each word using only the current epoch's data and the compressed multiscale priors, avoiding the need to store raw past data.</li>
</ul>
</li><br>

<li><b>Step 6: M-step — Hyperparameter Updates</b><br>
<ul>
<li><b>Description</b>: The hyperparameters for topic proportions and timescale weights are updated using maximum a posteriori (MAP) estimation.</li>
<li><b>Reference</b>: Section 3.2, Equations (9)–(11).</li>
<li><b>Purpose</b>: Allows the model to automatically learn how much weight to give each timescale for optimal temporal smoothing.</li>
</ul>
</li><br>

<li><b>Step 7: Updating Multiscale Distributions</b><br>
<ul>
<li><b>Description</b>: The multiscale word distributions are updated using exponential moving averages combining the newly inferred word distributions and the previous multiscale estimates.</li>
<li><b>Reference</b>: Section 3.2, Equations (5)–(6); Algorithm 1.</li>
<li><b>Purpose</b>: Prepares the time-smoothed memory distributions for use in the next discrete epoch.</li>
</ul>
</li>
</ul>

<b>Final Summary Sentence</b><br>
This paper solves the problem of modelling how topics in a streaming document collection evolve over time at multiple temporal granularities, and the authors claim their approach is better than existing alternatives (like DTM) because it performs fully online learning (no storage of past texts) while capturing both long-term trends and short-term bursts using multiscale priors, resulting in lower predictive perplexity and far less computational overhead.
