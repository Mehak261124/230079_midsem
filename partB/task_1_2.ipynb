<b>Assumption 1</b><br>
<b>Assumption:</b> Smooth temporal evolution of topics via exponential decay at multiple scales.<br>
<b>Why the method needs it:</b> The temporal linkage algorithm explicitly models future topics as exponential moving averages of past topics; if topics don't smoothly evolve, the Dirichlet prior points inference in the wrong direction.<br>
<b>Violation scenario:</b> A sudden, unprecedented event (like a new global pandemic) that introduces entirely new vocabulary instantaneously, making historical smoothed distributions irrelevant.<br>
<b>Paper reference:</b> Section 3.1, Equations (5)–(6) (Multiscale distribution update).<br><br>

<b>Assumption 2</b><br>
<b>Assumption:</b> The number of topics is fixed across all epochs.<br>
<b>Why the method needs it:</b> The multiscale temporal vectors are explicitly tracked for exactly K fixed slots; allowing K to vary would break the correspondence of history for these specific topics.<br>
<b>Violation scenario:</b> A multi-year social media corpus where completely new discourse communities are born and old ones die, requiring dynamic addition and removal of topic categories.<br>
<b>Paper reference:</b> Section 3.1 (Generative Model), parameter K.<br><br>

<b>Assumption 3</b><br>
<b>Assumption:</b> Documents within an epoch are completely exchangeable (Bag-of-Epochs assumption).<br>
<b>Why the method needs it:</b> The collapsed Gibbs sampling updates compute aggregate word-topic counts over the entire epoch, ignoring any temporal or sequential ordering of documents within that discrete time slice.<br>
<b>Violation scenario:</b> Intraday financial news where the epoch is 'one day', but market-moving events cause drastic mid-day shifts that the aggregate epoch bag-of-words completely averages out.<br>
<b>Paper reference:</b> Section 3.2, Equations (7)–(8) (Gibbs Sampling count statistics).
