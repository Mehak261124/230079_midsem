<b>Baseline or prior method compared against:</b><br>
Dynamic Topic Model (DTM) by Blei and Lafferty (2006).<br><br>

<b>Limitation of this baseline:</b><br>
The DTM captures topic evolution at only a single fixed timescale via a Gaussian random walk, preventing it from modeling complex real-world data where some vocabulary persists long-term while other vocabulary spikes sharply. Furthermore, DTM requires batch variational inference over the entire document collection, making it unscalable for continuous online data streams.<br><br>

<b>How the proposed method attempts to overcome this limitation:</b><br>
MDTM tracks multiple exponentially smoothed distributions at varying decay rates (timescales), constructing a composite Dirichlet prior that balances short-term bursts and long-term trends. It leverages a stochastic EM approach with collapsed Gibbs sampling that only requires data from the current epoch, entirely eliminating the need to store or re-process past texts.<br><br>

<b>Condition where the method would not outperform the baseline:</b><br>
A dataset with a small, fixed number of time epochs where topics evolve at a single, consistent, slow rate and the entire dataset fits easily in memory. In this scenario, DTM's batch variational inference uses all available data jointly to achieve a richer posterior estimate, whereas MDTM's "forgetful" lossy compression of past epochs via exponential smoothing provides no multiscale benefit and loses fine-grained historical information.
